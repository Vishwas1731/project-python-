```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b
```

```python
def reverse_string(text):
    return text[::-1]

def is_palindrome(text):
    clean = text.lower()
    return clean == clean[::-1]

def capitalize_words(text):
    return text.title()
```

```python
import subprocess
import json
import pandas as pd
import matplotlib.pyplot as plt
import os

# Create static folder
os.makedirs("static", exist_ok=True)

print("Running tests with coverage...")

# Run pytest with coverage
subprocess.run(
    "coverage run -m pytest",
    shell=True,
    check=True
)

# Generate JSON report
subprocess.run(
    "coverage json -o coverage_data.json",
    shell=True,
    check=True
)

print("Coverage JSON generated!")

# Check file exists
if not os.path.exists("coverage_data.json"):
    raise FileNotFoundError("coverage_data.json was not created")

# Load coverage data
with open("coverage_data.json", "r") as f:
    data = json.load(f)

files = data["files"]

coverage_summary = []

for file_name, file_data in files.items():

    summary = file_data["summary"]

    coverage_summary.append({
        "file": file_name,
        "coverage": summary["percent_covered"]
    })

# Convert to DataFrame
df = pd.DataFrame(coverage_summary)

print(df)

# Plot chart
plt.figure(figsize=(10, 5))

plt.bar(df["file"], df["coverage"])

plt.xticks(rotation=45, ha="right")

plt.ylabel("Coverage %")
plt.xlabel("Files")
plt.title("Test Coverage by File")

plt.tight_layout()

# Save chart
chart_path = "static/coverage_chart.png"

plt.savefig(chart_path)

print(f"Chart saved at: {chart_path}")
```

```python
from flask import Flask, render_template
import json

app = Flask(__name__)

@app.route("/")
def dashboard():
    with open("coverage_data.json", "r") as f:
        data = json.load(f)

    files = []

    for file_name, file_data in data["files"].items():
        summary = file_data["summary"]

        files.append({
            "name": file_name,
            "coverage": summary["percent_covered"],
            "missing_lines": file_data["missing_lines"]
        })

    uncovered_files = [
        f for f in files if f["coverage"] < 100
    ]

    return render_template(
        "index.html",
        files=files,
        uncovered_files=uncovered_files
    )

if __name__ == "__main__":
    app.run(debug=True)
```
